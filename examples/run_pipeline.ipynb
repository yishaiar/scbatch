## Create pipeline input

The pipeline snapshots adata.X after every matrix-transforming stage. The default QC rule removes only empty profiles.

```python
import numpy as np
from anndata import AnnData

X = np.arange(1, 121, dtype=float).reshape(30, 4)

adata = AnnData(X=X)
adata.obs["batch"] = ["reference"] * 15 + ["query"] * 15
adata.obs["is_anchor"] = [True] * 10 + [False] * 20
# adata.X
```

## Extract individual samples

`iter_samples` yields each original sample as an `n × m` NumPy array, followed by its input batch index and sample index.

```python
import sys
from pathlib import Path
import pandas as pd

worktree_root = Path(
    "/Users/yishai/Desktop/build_pip_package/scbatch-basic-structure-creation"
).resolve()
if str(worktree_root) in sys.path:
    sys.path.remove(str(worktree_root))
sys.path.insert(0, str(worktree_root))
for module_name in tuple(sys.modules):
    if module_name == "scbatch" or module_name.startswith("scbatch."):
        sys.modules.pop(module_name)


from scbatch.datasets import load_demo, summarize_samples
from scbatch.pp import filter_cells


# Keep every summary column visible when the notebook renders a wide table.
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", None)


adata = load_demo("synthetic")
adata_summary = summarize_samples(adata)
print(adata)
display(adata_summary)


filter_cells(adata, min_counts=1)
print(f"Observations after sample-wise filtering: {adata.n_obs}")
```

## Run the scbatch pipeline

ArcSinh, QQ correction, and core normalization currently use a temporary adata.X + 1 operation. This notebook demonstrates orchestration and representation storage, not validated batch-correction science.

```python
pipeline_result = run_pipeline(
    adata,
    method="qq",
    clustering_is_leiden=True,
)
print(pipeline_result.layers)
expected_layers = {
    "Gated",
    "arcsinh",
    "batch_corrected",
    "core_normalized",
    "scaled",
    "umap",
    "leiden",
}
assert expected_layers <= set(pipeline_result.layers)
assert "X_umap" in pipeline_result.obsm
assert "leiden" in pipeline_result.obs
pipeline_result
```
